```python
## Syntax
lambda argument : expression
```

```python
addition = lambda a, b : a + b
print(type(addition))
print(addition(5, 3))
```

```python
is_even = lambda x : x % 2 == 0
print(is_even(2))
```

```python
addition1 = lambda x, y, z: x + y + z
print(addition1(1, 2, 3))
```

```python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x : x ** 2, numbers))
print(numbers)
print(squares)
```
