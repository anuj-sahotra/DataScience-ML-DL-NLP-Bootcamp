```python
##Function with multple parameter

def add(a, b):
    """Add two numbers"""
    
    return a + b
print(add(4,5))
```

```python
## Default Parameter

def greet(name = "Champion"):
    print(f"Good Morning, {name}")
    
greet()
greet("Anuj")    
```

```python
### Varibale Length argument

# *args -> Positonal Argument
def print_num(*args):
    for numbers in args:
        print(numbers)
        
print_num(1, 2, 3, 4, 5)        
```

```python
# **kwargs-> Keyword argument
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")
        
print_details(name = "Champion", age = 23, City = "Kangra")        
```
