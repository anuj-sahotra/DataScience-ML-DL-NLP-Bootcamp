## for LOOP

```python
for i in range(1, 6):  # range(start, stop, step)
    print(i)
```

```python
# Odd numbers
for i in range(1, 11, 2):
    print(i)
```

```python
# even numbers
for i in range(2, 11, 2):
    print(i)
```

```python
for i in range(10, 0, -1):
    print(i, end=" ")
```

```python
# iterate over a string
char = "Python"
for i in char:
    print(i)
```

## while Loop

```python
# The while loop is used to execute a block of code as long as a condition is true.

# Initiatlize a variable with 0
count = 0
while count <= 5:
    print(count, end=" ")
    count += 1

print("\n")
 
# Initialize a variable with 1   
i = 1
while i <= 5:
    print(i, end=" ")
    i += 1    
```

## Loop Control Statement

```python
# break statement is used to exit a loop when a specific condition is met.

for i in range(1, 11):
    if i == 5:
        break
    print(i, end=" ")
```

```python
## continue statement is used to skip the current iteration of a loop and move on to the next iteration.

for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i, end=" ")
```

```python
## pass statement is used as a placeholder for future code. It does nothing when executed.
for i in range(10):
    pass
```

## Nested Loops

```python
for i in range(3):
    for j in range(2):
        print(f"i : {i} and j : {j}")
    
```

### Calculate the sum of first n natural numbers using a while and for loop

```python
# while loop

n = int(input("Range of Natural No. : "))

i = 1
result = 0
while i <= n:
    result += i
    i += 1
print(f"Sum of numbers from 1 to {n} : {result}")    
```

```python
# for loop

n = int(input("Range of Natural No. : "))

result = 0
for i in range(1, n + 1):
    result += i
    
print(f"Sum of numbers from 1 to {n} : {result}")    
```

### Prime numbers between 1 to 100

```python

for num in range(1, 101):
    if num > 1:
        for i in range(2, num):
            if num % i == 0:
                break
        else:
            print(num, end=" ")
        
```
