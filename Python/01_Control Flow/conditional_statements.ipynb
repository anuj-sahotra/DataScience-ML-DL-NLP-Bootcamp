```python
age = 20

if age >= 18:
    print("You are allowed to vote in the election.")
```

## else

```python
# The else statement executes a block of code if the codition in the if statement is False.

age = 16

if age >= 18:
    print("Eligible")
else:
    print("You are not eligible.")    
    
```

## elif 

```python
# The elif statement allows you to check multiple conditions. It stands for "else if"

age = 17
if age < 13:
    print("You are child.")
elif age < 18:
    print("You are a teenager.")
else:
    print("Adult")        
```

## Nested conditional statement

```python
# You can place one or more if, elif, or else statements inside another if, elif or else statement.

num = int(input("Enter the number : "))

if num >= 0:
    print("Posiitve number.")
    if num % 2 == 0:
        print(f"{num} is Even.")
    else:
        print(f"{num} is Odd.")
else:
    print("Negative number.")            
```

## Practical Example

```python
# Determine if he yeaer is a leap year 
year = int(input("Enter the year : "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a Leap Year")
else:
    print(f"{year} is not a leap year ")       
```
