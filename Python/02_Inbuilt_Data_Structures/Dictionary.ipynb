## Dictionary

- Key Value pair
- Unordered 
- unique and immutable

```python
dict = {}
print(type(dict))
```

```python
student = {
    "Name" : ["Anuj", "Champion"],
    "Age" : [20, 23],
    "Grade" :["A", "S++"]
}
print(student)
print(type(student))
print(len(student))
```

#### ACCESSING ELEMENT

```python
# Accessing dictionary element 
print(student['Name'])

# Accessing using get() method
print(student.get('Name'))

# If key is not found in dictinory
print(student.get('city'))
print(student.get('city', "Not Available"))
```

#### Modifying dictionary elements

```python
print(student)
```

```python
# Update value for key
student["Name"] =["Anuj", "Champion", "Warrior"]
print(student)

# Add a new key value pair
student["Country"] = "India"
print(student)
```

```python
# delete key
del student['Grade']
print(student)
```

#### Dictionary methods

```python
# keys() -> to get all the keys
keys = student.keys()
print(keys)

# values() to get all the values
values = student.values()
print(values)

# items() -> to get all the key value pairs
items = student.items()
print(items)
```

#### Shallow copy

```python
student_copy = student
print(student)
print(student_copy)
```

```python
student["Name"] = "Champion"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() # shallow copy
print(student_copy1)
print(student)
```

```python
student["Name"] = "Warrior"
print(student)
print(student_copy1)
```

#### Iterating over dictionaries

```python
student = {
    "Name" : ["Anuj", "Champion"],
    "Age" : [20, 23],
    "Grade" :["A", "S++"]
}
```

```python
# Iterating over keys

for keys in student.keys():
    print(keys) 
```

```python
# Iterating over values
for values in student.values():
    print(values) 
```

```python
## Ieratae over key value pairs

for key, values in student.items():
    print(f"{key} : {values}") 
```

#### Dictionary comprehension

```python
squares = {x : x ** 2 for x in range(1, 6)}
print(squares)
```

## Practical example

```python
## Use a dictionary to count the frequency of elements in list

numbers = [1, 1, 2, 3, 4, 3, 4, 5, 6, 7, 4, 5, 6]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1
        
print(frequency)       
```
