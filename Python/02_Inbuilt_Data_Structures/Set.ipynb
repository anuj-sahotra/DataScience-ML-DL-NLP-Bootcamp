### Basic set operations

```python
set1 = {1, 2, 3, 4, 5}

## Add element
set1.add(100)
print(set1)
```

#### Remove Vs Discard
--- Both are use to remove element
- remove() -> Throw error if element not found
- discard() - > Don't throw error if elementn not found

```python
set1.remove(4)
print(set1)
```

```python
set1.discard(1)
print(set1)

set1.discard(1000)
print(set1)
```

```python
# pop method
set1.pop()
print(set1)
```

```python
set1.clear()
print(set1)
```

```python
#### Set Membership test
```

```python
set1 = {1, 2, 3, 4, 5}
print(3 in set1)
print(10 in set1)
```

#### Mathematical Operation

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 9}
```

```python
print(set1.union(set2))
```

```python
print(set1.intersection(set2))
```

```python
set1.intersection_update(set2)
print(set1)      
```

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}

print(set1.difference(set2))
```

```python
print(set1.symmetric_difference(set2))
```

#### Set methods

```python
set1 = {1, 2, 3}
set2 = {1, 2, 3, 4, 5, 6}

# is subset
print(set1.issubset(set2))

# is superset
print(set1.issuperset(set2))
```
